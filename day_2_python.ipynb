## 2. Data Types and Control Flow

Python has four core built-in collection types and each one is suited to a different job.

**list** is ordered and mutable. Use it when you need to keep things in sequence and change them — for example, a list of column names that you will append to or reorder.

**tuple** is ordered and immutable. Use it when the contents must not change — for example, the shape of an array `(891, 15)` or the return value of a split operation `(train, test)`.

**set** has no duplicates and no guaranteed order. Use it when you need fast membership testing or to remove duplicates — checking whether a column name already exists in a schema.

**dict** maps keys to values. Use it when you need to look something up by name — a label encoder that maps `'cat' -> 0`, `'dog' -> 1`, or a schema that maps column names to their data types.

Beyond the basics, two control-flow patterns appear frequently in modern Python code.

**Ternary expression** — a compact one-line `if/else`:
```
value = a if condition else b
```
This is not just shorter; it reads more naturally when used in the right place. For example, `shuffle = True if is_training else False`.

## 3. Functions: *args, **kwargs, and Keyword-Only Arguments

Python functions become much more flexible when you understand three patterns.

**`*args`** collects any number of extra positional arguments into a tuple. The function does not need to know in advance how many values will be passed. Inside the function, `args` is a normal tuple you can loop over.

**`**kwargs`** collects any number of extra keyword arguments into a dictionary. This is how you pass an open-ended set of configuration values — for example, all the hyperparameters for a model — without naming each one in the function signature.

**Bare `*` separator** — placing a single `*` in the parameter list (not `*args`) forces every argument that comes after it to be passed by name, never by position:
```python
def train(X, y, *, epochs=10, lr=0.001):
    ...
```
This means `train(X, y, 5)` will raise a `TypeError`. The caller must write `train(X, y, epochs=5)`. This design pattern is used throughout scikit-learn specifically to prevent silent bugs where arguments are passed in the wrong order.

**`functools.partial`** creates a new function from an existing one with some arguments pre-filled. This is useful when you want to configure a function once and pass it as a callback without the receiving code needing to know about the pre-filled arguments.

````python
# 3a. *args — accept any number of metric names.
# Write a function report_metrics that takes any number of string arguments
# and prints a numbered list of the metric names it received.

def report_metrics(*metrics):
    print("metrics to report")
    print(metrics)
    
report_metrics("accuracy", "recall", "f1_score")
report_metrics("accuracy", "precision", "recall", "f1")
````

## 4. Comprehensions and Functional Patterns

Transforming data is the most common operation in machine learning code. Given a list of raw values, normalise them. Given a list of labels, encode them as integers. Given a list of column names, filter out the ones with null values. Python gives you two main tools for this.

**Comprehensions** build a new collection by applying an expression to every item in an iterable, with an optional filter condition:
```python
normalised = [(x - min_v) / (max_v - min_v) for x in raw_values]
even_only  = [x for x in data if x % 2 == 0]
```
Dict and set comprehensions work the same way with `{}` instead of `[]`.

**Generator expressions** look like list comprehensions but use `()`. The difference is that a generator is lazy — it computes one value at a time and never stores the full result in memory. When you have millions of rows, this distinction is critical.

**`map(fn, iterable)`** applies a function to every element and returns an iterator. It is equivalent to a list comprehension but can be more readable when you already have a named function.

**`filter(fn, iterable)`** keeps only the elements for which the function returns `True`. Together, `map` and `filter` are the building blocks of functional data pipelines.

````python
# 4a. List comprehension — manual Min-Max normalisation.
# Normalise raw_values to the range [0, 1] using the formula:
# normalised = (x - min) / (max - min)
# This is exactly what sklearn's MinMaxScaler computes — do it manually first.

raw_values = [10, 25, 5, 40, 15, 30]
min_v = min(raw_values)
max_v = max(raw_values)

normalize = []
for X in raw_values:
    X = (X - min_v) / (max_v - min_v)
    normalize.append(round(X,3))

print(f"Traditional {normalize}")

# normalised = [...]
# print([round(v, 3) for v in normalised])
normalized = [round((x - min_v) / (max_v - min_v),3 )for x in raw_values ]
print(normalized)
# Expected: [0.143, 0.571, 0.0, 1.0, 0.286, 0.714]
````

````python
# 4b. Dict comprehension — build a null report from two parallel lists.
# Create a dict that maps each column name to True if it has any nulls, False otherwise.

columns    = ["age", "salary", "department", "score"]
null_counts = [3, 0, 1, 0]

has_nulls = {col: (count > 0) for col, count in zip(columns, null_counts)} # (age, 3)
print(has_nulls)
# Expected: {"age": True, "salary": False, "department": True, "score": False}
````

````python
# 4c. Generator expression — compute sum of squares for 10 million numbers.
# A list comprehension would allocate ~80 MB of memory.
# A generator expression uses a fixed ~112 bytes regardless of size.

# import sys

list_result = [x**2 for x in range(10_000_000)]
gen_result  = (x**2 for x in range(10_000_000))

print(f"List size     : {sys.getsizeof(list_result):>14,} bytes")
print(f"Generator size: {sys.getsizeof(gen_result):>14,} bytes")

# print(list_result[:5])
# print(gen_result[:5])
# Compute the sum using the generator (no full list ever exists in memory)
# total = sum(x**2 for x in range(10_000_000))
# print(f"Sum: {total:,}")
````

````python
# 4d. map() and filter() with lambda.
# labels is a list of class names. Use map to convert each to uppercase.
# Use filter to keep only the "cat" labels.
# Then use map to encode labels as integers using a label_map dict.

labels    = ["cat", "dog", "cat", "bird", "dog", "cat"]
label_map = {"cat": 0, "dog": 1, "bird": 2}

upper_labels   = list(map(lambda s: s.upper(), labels))
cats_only      = list(filter(lambda s:s == 'cat', labels))
encoded_labels = list(map(lambda s: label_map[s], labels))

print(upper_labels)
print(cats_only)
print(encoded_labels)
````

````python
# 4e. Timing comparison — for-loop vs comprehension vs map.
# Run all three on a list of 1 million integers and compare the wall-clock time.
# Observe which is fastest and by how much.

import time
data = list(range(1_000_000))

# --- for-loop ---
start = time.perf_counter()
result_loop = []
for x in data:
    result_loop.append(x * 2)
loop_time = time.perf_counter() - start

# --- list comprehension ---
start = time.perf_counter()
result_comp = [x * 2 for x in data]
comp_time = time.perf_counter() - start

# --- map ---
start = time.perf_counter()
result_map = list(map(lambda x: x * 2, data))
map_time = time.perf_counter() - start

print(f"for-loop      : {loop_time:.4f}s")
print(f"comprehension : {comp_time:.4f}s")
print(f"map()         : {map_time:.4f}s")
# What does this tell you about how to write data transformations in Python?
````

## 5. Object-Oriented Programming: Classes

A class is a blueprint for creating objects that combine data (attributes) and behaviour (methods) in one place. Almost all major Python libraries are built around classes — scikit-learn estimators, PyTorch modules, Keras layers, and pandas DataFrames are all class instances.

**`__init__`** is called when you create an instance: `obj = MyClass()`. It sets up the initial state by assigning values to `self.attribute_name`.


**`@property`** creates a computed attribute — one that looks like a regular attribute when accessed but actually runs a function. The advantage is that it is always up-to-date with the current state of the object. You access it without parentheses: `obj.summary`, not `obj.summary()`.

**`@classmethod`** is a method that receives the class itself as its first argument (by convention named `cls`) instead of an instance. The most common use is an alternative constructor — a second way to create an instance, for example from a dictionary instead of individual arguments.

**`@staticmethod`** is a regular function that lives inside the class for organisational reasons but does not need access to `self` or `cls`.


*Grouping data and behaviour together — the foundation of all ML libraries* &nbsp;

### Key Concepts

- **Class vs object:** a class is a **blueprint**; an object (instance) is one specific thing made from that blueprint
- **`__init__`** runs automatically when you create an instance: `Student('Alice', [80, 90])`
- **`self`** refers to the specific instance — use it to store (`self.name = name`) and access attributes
- **`__repr__`** controls what `print()` shows — make it informative and unambiguous
- **`@property`** creates a **computed attribute** — accessed like a variable but runs a function behind the scenes
- **Methods** are functions that belong to a class — they always take `self` as their first argument

> **Tags:** `class` · `__init__` · `self` · `__repr__` · `@property` · `method`

````python
# 5a. Build a ModelConfig class with __init__, __repr__, and __eq__.
#
# Attributes: model_name (str), learning_rate (float), epochs (int)
#
# __repr__ should return a string like:
#   ModelConfig('rf', lr=0.001, epochs=100)
#
# __eq__ should return True only when all three attributes match between two instances.
# Hint: check isinstance(other, ModelConfig) first to handle comparisons with other types.

class ModelConfig:
    '''Stores hyperparameters for one model configuration.'''

    def __init__(self, model_name: str, learning_rate: float, epochs: int) -> None:
        self.model_name    = model_name
        self.learning_rate = learning_rate
        self.epochs        = epochs

    def __repr__(self) -> str:
        # !r adds quotes around string values automatically: 'rf' not rf
        return f"ModelConfig({self.model_name!r}, lr={self.learning_rate}, epochs={self.epochs})"

    def __eq__(self, other: object) -> bool:
        # Without this, == compares memory addresses (always False for different instances)
        if not isinstance(other, ModelConfig):
            return NotImplemented
        return (self.model_name    == other.model_name and
                self.learning_rate == other.learning_rate and
                self.epochs        == other.epochs)

cfg1 = ModelConfig("rf",  0.001, 100)
cfg2 = ModelConfig("rf",  0.001, 100)
cfg3 = ModelConfig("svm", 0.01,   50)

print(cfg1)            # ModelConfig('rf', lr=0.001, epochs=100)
print(cfg1 == cfg2)    # True  — same values, different objects
print(cfg1 == cfg3)    # False
print(cfg1 is cfg2)    # False — they are distinct objects in memory
````

## 6. Type Hints and Docstrings

**Type hints** are annotations that describe what type each argument and return value should be. Python does not enforce them at runtime — you can still pass the wrong type and the code will run until it breaks for some other reason. Their real value is for tools and for humans.

Your IDE uses type hints to provide autocomplete and catch type errors before you run anything. When you hover over a function call in VS Code, you see the parameter types and the return type immediately. This makes reading unfamiliar code much faster.

The syntax is straightforward:
```python
def compute_accuracy(y_true: list[int], y_pred: list[int]) -> float:
    ...
```

For more complex cases, the `typing` module provides `Optional` (the value can be `X` or `None`), `Union` (the value can be one of several types), and `Tuple` (a tuple with specific element types).

**Docstrings** explain what a function does, what its arguments are, what it returns, and what exceptions it can raise. The Google style used here is the standard in most machine learning teams:

```python
def my_function(path: str) -> pd.DataFrame:
    """One-line summary of what this function does.

    Args:
        path: A description of this argument.

    Returns:
        A description of what is returned.

    Raises:
        ValueError: When and why this is raised.
    """
```

The most important habit: write the docstring before writing the implementation. It forces you to think clearly about the interface before you write the code.

````python
# 6b. Write a complete Google-style docstring for the function below.
# Include: a one-line summary, Args section, Returns section, Raises section,
# and an Example section showing the expected output.
# After writing it, call help(train_test_split_manual) to see it rendered.

def compute_accuracy(y_true: list[int], y_pred: list[int]) -> float:
    """Compute classification accuracy as the fraction of correct predictions.

    Args:
        y_true: Ground-truth integer class labels.
        y_pred: Predicted integer class labels. Must be the same length as y_true.

    Returns:
        Accuracy as a float between 0.0 (no correct predictions) and 1.0 (all correct).

    Raises:
        ValueError: If y_true and y_pred have different lengths.

    Example:
        >>> compute_accuracy([0, 1, 1, 0], [0, 1, 0, 0])
        0.75
    """
    if len(y_true) != len(y_pred):
        raise ValueError(f"Length mismatch: y_true has {len(y_true)}, y_pred has {len(y_pred)}")
    correct = sum(t == p for t, p in zip(y_true, y_pred))
    return correct / len(y_true)

print(compute_accuracy([0, 1, 1, 0], [0, 1, 0, 0]))
print()
help(compute_accuracy)
````

## 7. Exception Handling
Production code fails in ways you cannot always predict — a file path is wrong, a column is missing, the data shape does not match what the model expects. The difference between robust code and fragile code is how it handles these failures.

**`try / except / else / finally`**
- `try` — the code that might fail
- `except ExceptionType as e` — what to do when it fails
- `else` — runs only if no exception was raised (useful for success-path logic)
- `finally` — always runs, even if an exception was raised (use for cleanup)

**Custom exceptions** — subclass a built-in exception to give callers something specific to catch. A function that raises `DataError` is much clearer than one that raises a generic `Exception`.

**`raise X from e`** — this pattern wraps the original exception inside a new one. The caller sees your custom message, but the full original traceback is preserved. Always use this instead of bare `raise DataError(msg)` when you are converting one exception type to another.



````python
# 7a. Define a custom DataError exception.
# It should inherit from ValueError.
# Test it by raising it inside a try/except block and printing the message.

class DataError(ValueError):
    '''Raised for any invalid Dataset operation.'''
    pass

# Test: raise it inside try/except
try:
    raise DataError("Column 'target' not found. Available: ['age', 'salary']")
except DataError as e:
    print(f"Caught DataError: {e}")
````

````python
# try / except / else / finally
# Teaching note: walk through each block one at a time.
# else  — runs ONLY if no exception was raised in try
# finally — ALWAYS runs, even if there is a return in try

def safe_load(path: str) -> pd.DataFrame:
    """Load a CSV file, raising DataError on any failure.

    Args:
        path: Path to the CSV file.

    Returns:
        A pandas DataFrame.

    Raises:
        DataError: If the file is not found, unparseable, or empty.
    """
    try:
        df = pd.read_csv(path)
        if df.empty:
            raise DataError(f"'{path}' loaded successfully but contains no rows.")
    except FileNotFoundError as e:
        # raise X from e: attaches the original traceback to the new exception
        # The caller sees "DataError: File not found" AND the original FileNotFoundError
        raise DataError(f"File not found: '{path}'") from e
    except pd.errors.ParserError as e:
        raise DataError(f"Could not parse '{path}' as a CSV file.") from e
    else:
        # This block runs only when no exception was raised
        print(f"Loaded {len(df)} rows and {len(df.columns)} columns successfully.")
        return df
    finally:
        # This always runs — logging, cleanup, connection teardown
        print(f"Load attempt complete for: '{path}'")

# Test the error path
try:
    safe_load("/home/fm-pc-lt-269/Desktop/Sagarmatha /data/titanic.csv")
except DataError as e:
    print(f"Caught: {e}")
````
